```python
import mne
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import sys
import pyedflib
import h5py
from mne.io.edf.edf import _read_annotations_edf

%matplotlib qt
```

```python
def edf_to_h5(edf_file):
    edf_directory = os.path.dirname(edf_file)
    base_name = os.path.splitext(os.path.basename(edf_file))[0]
    
    h5_file = os.path.join(edf_directory, f"{base_name}.h5")
    
    try:
        raw = mne.io.read_raw_edf(edf_file, preload=True)

        data, times = raw[:, :]
        
        with h5py.File(h5_file, 'w') as h5f:
            # Store global metadata
            h5f.attrs['start_datetime'] = raw.info['meas_date'].isoformat() if raw.info['meas_date'] else 'unknown'
            h5f.attrs['sfreq'] = raw.info['sfreq']
            h5f.attrs['ch_names'] = np.array(raw.info['ch_names'], dtype='S')
            
            h5f.create_dataset('data', data=data, dtype=data.dtype)
            h5f.create_dataset('times', data=times, dtype=times.dtype)
            
            if raw.annotations:
                annot_group = h5f.create_group('annotations')
                annot_group.create_dataset('onset', data=raw.annotations.onset, dtype=raw.annotations.onset.dtype)
                annot_group.create_dataset('duration', data=raw.annotations.duration, dtype=raw.annotations.duration.dtype)
                annot_group.create_dataset('description', data=np.array(raw.annotations.description, dtype='S'))

    except Exception as e:
        print(f"Error processing {edf_file}: {e}")
```

```python
annotations = []
for root, dirs, files in os.walk('data/source/ds003876'):
    for file in files:
        if file.endswith('.edf') and 'sleep' not in file:
            file_path = os.path.join(root, file)
            edf_to_h5(file_path)


```
