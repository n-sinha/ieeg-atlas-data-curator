```python
import mne
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import sys
import pyedflib
import h5py

%matplotlib qt
```

```python
def edf_to_h5(edf_file):
    edf_directory = os.path.dirname(edf_file)
    base_name = os.path.splitext(os.path.basename(edf_file))[0]
    
    h5_file = os.path.join(edf_directory, f"{base_name}.h5")
    
    # Open the EDF file
    with pyedflib.EdfReader(edf_file) as edf:
        n_signals = edf.signals_in_file
        signal_labels = edf.getSignalLabels()
        signal_freqs = edf.getSampleFrequencies()
        
        # Create an HDF5 file
        with h5py.File(h5_file, 'w') as h5f:
            # Store global metadata
            h5f.attrs['start_datetime'] = edf.getStartdatetime().isoformat()
            h5f.attrs['patient_info'] = edf.getPatientAdditional()
            h5f.attrs['recording_info'] = edf.getRecordingAdditional()
            
            # Create a group for each signal
            for i in range(n_signals):
                # Read the signal data
                signal_data = edf.readSignal(i)
                dset = h5f.create_dataset(signal_labels[i], data=signal_data, dtype='float64')
                
                # Add attributes for metadata
                dset.attrs['sample_frequency'] = signal_freqs[i]
                dset.attrs['label'] = signal_labels[i]
            
            annotations = edf.readAnnotations()
            if annotations:
                annot_group = h5f.create_group('annotations')
                annot_group.create_dataset('onset', data=annotations[0])
                annot_group.create_dataset('duration', data=annotations[1])
                annot_group.create_dataset('description', data=np.array(annotations[2], dtype='S'))
```

```python
annotations = []
for root, dirs, files in os.walk('data/source/ds003876'):
    for file in files:
        if file.endswith('.edf'):
            edf_file_path = os.path.join(root, file)
            edf_to_h5(edf_file_path)


```
